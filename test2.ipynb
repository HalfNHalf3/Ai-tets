```python
import numpy as np

def calculator( num1, num2, operation):
  if operation == '1':
    return num1 + num2
  if operation == '2':
    return num1 - num2
  if operation == '3':
    return num1 * num2
  if operation == '4':
    return num1 / num2
  else:
    return "Invalid operation"

def randomNumberGenerator():
  return np.random.randint(100)

def rockPaperScissors():
  throw = np.random.randint(3)
  if throw == 0:
    return throw#rock
  if throw == 1:
    return throw#paper
  if throw == 2:
    return throw#scissors

answer = " "
option = int(input("Doyou wish to use the calculator or play rock paper scissors:\n 1:Calculator\n 2:Rock Paper Scissors \n"))
if option == 1:
  option1 = int(input("Do you want to use you own numbers: \n 1:Yes \n 2:No\n"))
  if option1 == 1:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    operation = input("Enter operation:\n 1:Addition\n 2:Subtraction \n 3:Multiplication \n 4:Division \n")
  elif option1 == 2:
    num1 = randomNumberGenerator()
    num2 = randomNumberGenerator()
    print("First number: ", num1)
    print("Second number: ", num2)
    operation = input("Enter operation:\n 1:Addition\n 2:Subtraction \n 3:Multiplication \n 4:Division \n")

  else:
    answer = print("Invalid option")
  if answer == " ":
    result = calculator(num1, num2, operation)
    print(result)
  else:
    print("Invalid inputs")
if option == 2:
  userThrow = int(input("Enter your throw: \n 0:Rock\n 1:Paper\n 2:Scissors\n"))
  computerThrow = rockPaperScissors()
  if userThrow == 0 and computerThrow == 1:
    print("Computer wins\n")
    print("Computer threw Paper")
  elif userThrow == 0 and computerThrow == 2:
    print("User wins\n")
    print("Computer threw Scissors")
  elif userThrow == 1 and computerThrow == 0:
    print("User wins")
    print("Computer threw Rock")
  elif userThrow == 1 and computerThrow == 2:
    print("Computer wins")
    print("Computer threw Scissors")
  elif userThrow == 2 and computerThrow == 0:
    print("Computer wins")
    print("Computer threw Rock")
  elif userThrow == 2 and computerThrow == 1:
    print("User wins")
    print("Computer threw Paper")
  else:
    print("Draw")
```
